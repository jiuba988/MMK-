```python

from google import genai
from google.colab import userdata
from google.genai.errors import APIError
import time

# 1. 安全读取 API Key
gemini_key = userdata.get("GEMINI_API_KEY")

# 2. 创建 Gemini 客户端
client = genai.Client(api_key=gemini_key)

# 3. 指定模型
MODEL_NAME = "gemini-3.5-flash-lite"

# 4. 测试内容
prompt = """
你现在是 Myanmar Rate AI 缅甸汇率咨询专员。

请分别用以下三种语言打招呼：
1. 中文
2. 缅甸语
3. 英文

告诉用户你可以帮助查询 MMK、THB、USD、CNY 汇率。

现在只是测试语言能力，不要编造任何汇率。
"""

# 5. 自动重试
for attempt in range(4):

    try:
        print(f"正在测试 {MODEL_NAME}...")
        print(f"第 {attempt + 1} 次请求\n")

        response = client.models.generate_content(
            model=MODEL_NAME,
            contents=prompt
        )

        print("✅ Gemini 模型测试成功！")
        print("当前模型：", MODEL_NAME)
        print("\n", response.text)

        break

    except APIError as e:

        print("API 错误代码：", e.code)

        if e.code in [503, 429] and attempt < 3:

            wait_time = 10 * (attempt + 1)

            print(f"⚠️ 服务暂时繁忙，等待 {wait_time} 秒后重试...")
            time.sleep(wait_time)

        else:

            print("❌ Gemini 请求失败：", str(e))
            break

    except Exception as e:

        print("❌ 发生未知错误：", repr(e))
        break
```

```python

import requests
import pandas as pd
from io import StringIO

url = "https://forex.cbm.gov.mm/index.php/fxrate"

response = requests.get(
    url,
    headers={"User-Agent": "Mozilla/5.0"},
    timeout=30
)
response.raise_for_status()

tables = pd.read_html(StringIO(response.text))

# 第一张表：USD 官方参考价和市场交易参考价
usd_table = tables[0]
usd = usd_table[usd_table["Currency.1"] == "USD"].iloc[0]

# 第二张表：CNY 和 THB 账户交易参考价
cross_table = tables[1]

result = {}

for currency in ["CNY", "THB"]:
    row = cross_table[
        cross_table["Currency.1"] == currency
    ].iloc[0]

    result[currency] = {
        "account_rate_1": float(row.iloc[3]),
        "account_rate_2": float(row.iloc[4])
    }

result["USD"] = {
    "reference_rate": float(usd["Reference Rate"]),
    "market_trading_rate": float(usd["Market Trading Rate"])
}

print("🇲🇲 CBM 汇率提取结果")
print(result)
```

```python
import asyncio
import requests
import pandas as pd
import re

from io import StringIO
from datetime import datetime
from zoneinfo import ZoneInfo

from telegram import Update
from telegram.ext import (
    Application,
    CommandHandler,
    MessageHandler,
    ContextTypes,
    filters
)

from google import genai
from google.colab import userdata
from google.genai.errors import APIError

# ==============================
# 1. 读取密钥
# ==============================
TELEGRAM_TOKEN = userdata.get("TELEGRAM_BOT_TOKEN")
GEMINI_KEY = userdata.get("GEMINI_API_KEY")

client = genai.Client(api_key=GEMINI_KEY)
MODEL_NAME = "gemini-3.5-flash-lite"

# ✨ 打磨点 1：完整补齐缅甸央行所有公开交易外币的对应国旗图标，绝不使用 🌐 敷衍
FLAG_MAP = {
    "USD": "🇺🇸", "EUR": "🇪🇺", "SGD": "🇸🇬", "GBP": "🇬🇧", "CHF": "🇨🇭",
    "JPY": "🇯🇵", "AUD": "🇦🇺", "BDT": "🇧🇩", "BND": "🇧🇳", "CAD": "🇨🇦",
    "DKK": "🇩🇰", "HKD": "🇭🇰", "INR": "🇮🇳", "IDR": "🇮🇩", "KRW": "🇰🇷",
    "LAK": "🇱🇦", "MYR": "🇲🇾", "NZD": "🇳🇿", "NOK": "🇳🇴", "PKR": "🇵🇰",
    "PHP": "🇵🇭", "RUB": "🇷🇺", "SAR": "🇸🇦", "SEK": "🇸🇪", "LKR": "🇱🇰",
    "VND": "🇻🇳", "KWD": "🇰🇼", "NPR": "🇳🇵", "KES": "🇰🇪", "EGP": "🇪🇬",
    "KHR": "🇰🇭", "CNY": "🇨🇳", "THB": "🇹🇭",
    "BRL": "🇧🇷", "CZK": "🇨🇿", "ILS": "🇮🇱", "RSD": "🇷🇸", "ZAR": "🇿🇦"
}

# ==============================
# 2. Gemini 聊天设置
# ==============================
SYSTEM_PROMPT = """
You are Myanmar Rate AI. Your only job is to format and translate the structured conversion results and exchange rates provided by Python into a natural response in the user's language (Burmese, Chinese, or English).

Rules:
1. Reply in the same language used by the user.
2. Do NOT perform any math or calculation yourself. Only use the calculations provided by the Python runtime.
3. Do NOT invent or alter any numbers. Keep all numbers exactly as calculated by Python.
4. Present the numbers in a very clean, structured layout using emojis, bolding key rates.
"""

# ==============================
# 3. 带重试机制的网络数据爬取
# ==============================
def get_cbm_rates(retries=3, delay=3):
    """✨ 打磨点 3：增加带延时自动重试的鲁棒网络请求机制，防止缅甸央行网站偶发网络抖动导致机器人瘫痪"""
    url = "https://forex.cbm.gov.mm/index.php/fxrate"
    headers = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36"}

    for i in range(retries):
        try:
            response = requests.get(url, headers=headers, timeout=15)
            response.raise_for_status()
            tables = pd.read_html(StringIO(response.text))

            # 安全解析
            usd_table = tables[0]
            usd_row = usd_table[usd_table["Currency.1"] == "USD"].iloc[0]
            cross_table = tables[1]

            result = {
                "USD": {
                    "reference_rate": float(str(usd_row["Reference Rate"]).replace(",", "")),
                    "market_trading_rate": float(str(usd_row["Market Trading Rate"]).replace(",", ""))
                }
            }

            for currency in ["CNY", "THB"]:
                row = cross_table[cross_table["Currency.1"] == currency].iloc[0]
                result[currency] = {
                    "account_rate_1": float(str(row.iloc[3]).replace(",", "")),
                    "account_rate_2": float(str(row.iloc[4]).replace(",", ""))
                }
            return result
        except Exception as e:
            if i == retries - 1:
                raise e
            import time
            time.sleep(delay)

# ==============================
# Python 高精度数据过滤器 & 计算器
# ==============================
def build_rate_reply(user_message, rates, fetch_time):
    msg = user_message.upper()

    if any(word in msg for word in ["人民币", "RMB", "CNY", "元"]):
        currency = "CNY"
    elif any(word in msg for word in ["泰铢", "THB", "铢"]):
        currency = "THB"
    elif any(word in msg for word in ["美元", "美金", "USD", "刀"]):
        currency = "USD"
    else:
        return None

    # ✨ 打磨点 2：优化正则清洗提取，支持带千分位、空格或中文字符的各种非标输入识别
    clean_msg = msg.replace(",", "")
    amount_match = re.search(r"(\d+(\.\d+)?)", clean_msg)
    amount = float(amount_match.group(1)) if amount_match else 1.0

    data = rates.get(currency, {})
    flag = FLAG_MAP.get(currency, "💵")

    if currency == "USD":
        rate1 = data.get("reference_rate")
        rate2 = data.get("market_trading_rate")
        label1 = "🏛️ 官方参考价"
        label2 = "🏪 市场交易价"
    else:
        rate1 = data.get("account_rate_1")
        rate2 = data.get("account_rate_2")
        label1 = "📈 官方参考价"
        label2 = "📉 市场参考价 B"

    if rate1 is None or rate2 is None:
        return "⚠️ 暂时无法获取完整汇率数据。"

    reply = f"""📊 *缅甸央行 (CBM) 即时计算结果*
━━━━━━━━━━━━━━━━━━
💱 兑换方向：{flag} *{currency}* ➔ 🇲🇲 *MMK*
🪙 兑换金额：*{amount:,.2f}* {currency}

{label1}：
➡️ *{amount * rate1:,.2f}* MMK  (汇率: {rate1:,.2f})

{label2}：
➡️ *{amount * rate2:,.2f}* MMK  (汇率: {rate2:,.2f})
━━━━━━━━━━━━━━━━━━
🕒 报价时间：{fetch_time}
"""
    return reply

# ==============================
# 4. 指令处理器
# ==============================
async def start(update: Update, context: ContextTypes.DEFAULT_TYPE):
    welcome = """👋 **您好！我是缅甸央行实时汇率查询机器人**

你可以点击使用下方快捷指令：
🌍 /rate - 获取核心货币 (USD, CNY, THB) 的官方参考与市场即时双价。
🗺️ /other - 查询除美、中、泰以外的所有其他国家官方与市场汇率。

🧮 **高精准快捷计算**：
您可以直接发送类似 `100美元`、`1500泰铢` 或 `500元`，我将自动帮您套用最新数据完成换算！"""
    await update.message.reply_text(welcome, parse_mode="Markdown")

async def rate(update: Update, context: ContextTypes.DEFAULT_TYPE):
    await update.message.reply_text("⏳ 正在读取 CBM 最新汇率，请稍等……")
    try:
        data = await asyncio.to_thread(get_cbm_rates)
        fetch_time = datetime.now(ZoneInfo("Asia/Bangkok")).strftime("%Y-%m-%d %H:%M:%S")

        message = f"""📈 **缅甸核心货币实时牌价**
━━━━━━━━━━━━━━━━━━
🇺🇸 **USD / MMK (美元)**
🏛️ 官方参考价：*{data["USD"]["reference_rate"]:,.2f} MMK*
🏪 市场交易价：*{data["USD"]["market_trading_rate"]:,.2f} MMK*
━━━━━━━━━━━━━━━━━━
🇨🇳 **CNY / MMK (人民币)**
🏛️ 官方参考价：*{data["CNY"]["account_rate_1"]:,.2f} MMK*
🏪 市场参考价：*{data["CNY"]["account_rate_2"]:,.2f} MMK*
━━━━━━━━━━━━━━━━━━
🇹🇭 **THB / MMK (泰铢)**
🏛️ 官方参考价：*{data["THB"]["account_rate_1"]:,.2f} MMK*
🏪 市场参考价：*{data["THB"]["account_rate_2"]:,.2f} MMK*
━━━━━━━━━━━━━━━━━━
📌 数据源：缅甸中央银行 (CBM)
🕒 抓取时间（曼谷时间）：{fetch_time}
"""
        await update.message.reply_text(message, parse_mode="Markdown")
    except Exception as e:
        await update.message.reply_text(f"❌ 获取 CBM 汇率失败，请稍后再试。\n错误信息：{str(e)[:150]}")

async def other(update: Update, context: ContextTypes.DEFAULT_TYPE):
    await update.message.reply_text("⏳ 正在获取其他国家的最新外币双价数据，请稍等……")
    url = "https://forex.cbm.gov.mm/index.php/fxrate"
    headers = {"User-Agent": "Mozilla/5.0"}

    try:
        response = await asyncio.to_thread(requests.get, url, headers=headers, timeout=30)
        tables = pd.read_html(StringIO(response.text))

        if len(tables) < 2:
            await update.message.reply_text("⚠️ 未在央行官网找到完整的其他外币交易表。")
            return

        cross_table = tables[1]
        cross_table.columns = [
            "Currency", "Currency.1", "Value",
            "Account_Rate_1", "Account_Rate_2"
        ]

        lines = ["🗺️ **缅甸央行 - 其他外币参考双价汇总**", "━━━━━━━━━━━━━━━━━━"]

        for _, row in cross_table.iterrows():
            name = str(row["Currency"]).strip()
            code = str(row["Currency.1"]).strip().upper()

            if code in ["USD", "CNY", "THB"]:
                continue

            flag = FLAG_MAP.get(code, "🌐")

            try:
                rate1 = float(str(row["Account_Rate_1"]).replace(",", ""))
                rate2 = float(str(row["Account_Rate_2"]).replace(",", ""))

                lines.append(f"{flag} *{code} ({name})* / MMK")
                lines.append(f"🏛️ 官方参考 A：*{rate1:,.2f}* MMK")
                lines.append(f"🏪 市场参考 B：*{rate2:,.2f}* MMK")
                lines.append("----------------------------------")
            except:
                continue

        lines.append("🕒 提示：由于汇率实时浮动，以上数据仅供账户交易和换算参考。")

        chunk = ""
        for line in lines:
            if len(chunk) + len(line) + 2 > 4000:
                await update.message.reply_text(chunk, parse_mode="Markdown")
                chunk = ""
            chunk += line + "\n"

        if chunk:
            await update.message.reply_text(chunk, parse_mode="Markdown")

    except Exception as e:
        await update.message.reply_text(f"❌ 抓取其他外币汇率时发生异常: {str(e)[:150]}")

async def chat(update: Update, context: ContextTypes.DEFAULT_TYPE):
    user_message = update.message.text
    fetch_time = datetime.now(ZoneInfo("Asia/Bangkok")).strftime("%Y-%m-%d %H:%M:%S")

    try:
        rates = await asyncio.to_thread(get_cbm_rates)
        rate_reply = build_rate_reply(user_message, rates, fetch_time)

        if rate_reply is not None:
            # 有计算命令时，通过 Gemini 润色排版
            try:
                prompt = f"""用户输入的消息: \"{user_message}\"
Python 计算出的精确汇率文本如下:
{rate_reply}

请严格基于 Python 提供的上述结果及数据进行润色和翻译。确保不更改计算出的任何数值，生成精美易读的回复。"""
                response = client.models.generate_content(
                    model=MODEL_NAME,
                    contents=prompt,
                    config={
                        "system_instruction": SYSTEM_PROMPT,
                        "temperature": 0.1
                    }
                )
                await update.message.reply_text(response.text, parse_mode="Markdown")
                return
            except Exception as e:
                await update.message.reply_text(rate_reply, parse_mode="Markdown")
                return
    except Exception as e:
        print(f"汇率获取或计算出错: {e}")

    # 普通聊天交给 Gemini
    try:
        prompt = f"用户发来的消息：\"{user_message}\"。请作为缅甸汇率小助手进行针对性的友好回复（不要瞎编实时数据，可提示其输入 /rate 快速查询）"
        response = client.models.generate_content(
            model=MODEL_NAME,
            contents=prompt,
            config={
                "system_instruction": SYSTEM_PROMPT,
                "temperature": 0.7
            }
        )
        await update.message.reply_text(response.text, parse_mode="Markdown")
    except Exception as e:
        await update.message.reply_text("💡 提示：输入 `/rate` 即可快速获取并折算实时外币汇率。")

# ==============================
# 5. 启动与安全实例注销逻辑
# ==============================
async def main():
    global app

    if 'app' in globals() and app is not None:
        print("🔄 检测到已在运行的机器人实例，正在安全释放端口与连接...")
        try:
            if app.updater and app.updater.running:
                await app.updater.stop()
            await app.stop()
            await app.shutdown()
            print("✅ 历史连接清理完成！")
        except Exception as err:
            print(f"⚠️ 释放旧连接时发生小错误（可忽略）: {err}")

    app = Application.builder().token(TELEGRAM_TOKEN).build()

    # 安全重置 webhook，规避冲突报错
    try:
        await app.bot.delete_webhook(drop_pending_updates=True)
    except Exception as e:
        print(f"⚠️ 清除旧 webhook 失败 (可忽略): {e}")

    # 注册路由处理器
    app.add_handler(CommandHandler("start", start))
    app.add_handler(CommandHandler("rate", rate))
    app.add_handler(CommandHandler("other", other))
    app.add_handler(MessageHandler(filters.TEXT & ~filters.COMMAND, chat))

    print("🚀 缅甸汇率助手 Telegram Bot 正在进行轮询监听中，请勿停止本单元格...")
    await app.initialize()
    await app.updater.start_polling(drop_pending_updates=True)
    await app.start()
    await asyncio.Event().wait()

if __name__ == "__main__":
    try:
        await main()
    except Exception as e:
        print(f"启动异常: {e}")
```
