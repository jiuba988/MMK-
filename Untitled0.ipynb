```python
# ============================================================
# Myanmar Rate AI | 单 Cell 完整版（修正版）
# CBM 官网实时抓取 + 国家按键选择 + 单币种回复 + 趋势比较 + 自动换算
# Gemini 多语言聊天 + 用户聊天风格 + Google Drive 历史保存
# 适用于 Google Colab：整段放进一个代码单元格运行
# ============================================================

# 固定 Colab 兼容版本：避免 -U 把 Colab / google-adk 依赖升级冲突
%pip -q install "python-telegram-bot>=21,<23" "google-genai>=2.12.1,<3" "requests==2.32.4" "google-auth==2.49.0" beautifulsoup4 lxml

import asyncio
import json
import os
import re
import logging
import hashlib
from decimal import Decimal, InvalidOperation
from datetime import datetime
from zoneinfo import ZoneInfo

import requests
from bs4 import BeautifulSoup
from google import genai
from google.genai import types
from google.colab import userdata, drive

from telegram import Update, InlineKeyboardButton, InlineKeyboardMarkup
from telegram.ext import (Application, CommandHandler, MessageHandler, CallbackQueryHandler, ContextTypes, filters)

# 1. 基础设置
TELEGRAM_TOKEN = userdata.get("TELEGRAM_BOT_TOKEN")
GEMINI_KEY = userdata.get("GEMINI_API_KEY")
if not TELEGRAM_TOKEN:
    raise ValueError("Colab Secrets 缺少 TELEGRAM_BOT_TOKEN")
if not GEMINI_KEY:
    raise ValueError("Colab Secrets 缺少 GEMINI_API_KEY")

client = genai.Client(api_key=GEMINI_KEY)
MODEL_NAME = "gemini-3.5-flash-lite"
TZ = ZoneInfo("Asia/Bangkok")
CBM_URL = "https://forex.cbm.gov.mm/index.php/fxrate"
CBM_API = "https://forex.cbm.gov.mm/api/latest"
HEADERS = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 Chrome/120 Safari/537.36"}

logging.basicConfig(level=logging.INFO, format="%(asctime)s | %(levelname)s | %(message)s")
logger = logging.getLogger("MyanmarRateAI")

# Drive 保存历史。若未能挂载，会退回临时目录（运行时重启后历史会丢失）。
try:
    if not os.path.isdir("/content/drive/MyDrive"):
        drive.mount("/content/drive")
    DATA_DIR = "/content/drive/MyDrive/MyanmarRateAI"
    os.makedirs(DATA_DIR, exist_ok=True)
    print("✅ Google Drive 已连接：历史汇率和聊天偏好可跨运行时保存")
except Exception as e:
    DATA_DIR = "/content/mmk_rate_bot_data"
    os.makedirs(DATA_DIR, exist_ok=True)
    print("⚠️ Google Drive 未连接，将使用临时存储；Colab 重启后历史可能丢失。", e)

HISTORY_FILE = os.path.join(DATA_DIR, "cbm_rate_history.json")
PROFILE_FILE = os.path.join(DATA_DIR, "mmk_user_profiles.json")

# 2. 货币与语言
FLAG_MAP = {
    "USD":"🇺🇸", "EUR":"🇪🇺", "SGD":"🇸🇬", "GBP":"🇬🇧", "CHF":"🇨🇭",
    "JPY":"🇯🇵", "AUD":"🇦🇺", "BDT":"🇧🇩", "BND":"🇧🇳", "CAD":"🇨🇦",
    "DKK":"🇩🇰", "HKD":"🇭🇰", "INR":"🇮🇳", "IDR":"🇮🇩", "KRW":"🇰🇷",
    "LAK":"🇱🇦", "MYR":"🇲🇾", "NZD":"🇳🇿", "NOK":"🇳🇴", "PKR":"🇵🇰",
    "PHP":"🇵🇭", "RUB":"🇷🇺", "SAR":"🇸🇦", "SEK":"🇸🇪", "LKR":"🇱🇰",
    "VND":"🇻🇳", "KWD":"🇰🇼", "NPR":"🇳🇵", "KES":"🇰🇪", "EGP":"🇪🇬",
    "KHR":"🇰🇭", "CNY":"🇨🇳", "THB":"🇹🇭", "BRL":"🇧🇷", "CZK":"🇨🇿",
    "ILS":"🇮🇱", "RSD":"🇷🇸", "ZAR":"🇿🇦", "TWD":"🇹🇼", "AED":"🇦🇪"
}
CORE_CURRENCIES = ["USD", "CNY", "THB", "MYR", "SGD", "TWD", "JPY", "KRW", "AED", "LKR"]
CURRENCY_KEYWORDS = {
    "CNY":["人民币","RMB","CNY","元","YUAN","တရုတ်ယွမ်","ယွမ်"],
    "THB":["泰铢","THB","铢","BAHT","บาท","ဘတ်"],
    "USD":["美元","美金","USD","刀","DOLLAR","DOLLARS","ဒေါ်လာ"],
    "MYR":["马币","马来西亚","MYR","RINGGIT","ရင်းဂစ်"],
    "SGD":["新币","新加坡","SGD","SINGAPORE DOLLAR"],
    "TWD":["台币","台湾","TWD"], "JPY":["日元","日本","JPY","YEN"],
    "KRW":["韩元","韩国","KRW"], "AED":["迪拉姆","迪拜","阿联酋","AED"],
    "LKR":["斯里兰卡","LKR"], "VND":["越南盾","越南","VND"],
    "EUR":["欧元","EUR"], "GBP":["英镑","GBP"], "AUD":["澳元","AUD"],
    "INR":["印度卢比","INR"], "IDR":["印尼盾","IDR"], "HKD":["港币","HKD"],
    "PHP":["菲律宾比索","PHP"], "LAK":["老挝基普","LAK"],
    "BDT":["孟加拉塔卡","BDT"], "NPR":["尼泊尔卢比","NPR"],
    "KHR":["柬埔寨瑞尔","KHR"]
}
MYANMAR_DIGITS = {"၀":"0","၁":"1","၂":"2","၃":"3","၄":"4","၅":"5","၆":"6","၇":"7","၈":"8","၉":"9"}

def translate_myanmar_digits(text):
    for old, new in MYANMAR_DIGITS.items():
        text = str(text).replace(old, new)
    return text

def detect_language(text):
    if re.search(r"[\u1000-\u109F]", text):
        return "my"
    if re.search(r"[\u4e00-\u9fff]", text):
        return "zh"
    return "en"

def identify_currency(text, available_codes=None):
    normalized = translate_myanmar_digits(text).upper()
    for code, words in CURRENCY_KEYWORDS.items():
        for word in sorted(words, key=len, reverse=True):
            if word.upper() in normalized:
                return code
    codes = set(FLAG_MAP)
    if available_codes:
        codes.update(available_codes)
    for code in sorted(codes, key=len, reverse=True):
        if re.search(rf"(?<![A-Z]){re.escape(code)}(?![A-Z])", normalized):
            return code
    return None

def to_script_bold_text(text):
    result = []
    for ch in str(text):
        if "A" <= ch <= "Z":
            result.append(chr(0x1D468 + ord(ch) - ord("A")))
        elif "a" <= ch <= "z":
            result.append(chr(0x1D482 + ord(ch) - ord("a")))
        else:
            result.append(ch)
    return "".join(result)

# 3. 用户聊天风格（不猜测用户性别）
def load_json_file(path, default):
    try:
        if os.path.exists(path):
            with open(path, "r", encoding="utf-8") as f:
                return json.load(f)
    except Exception as e:
        logger.warning("读取 JSON 失败 %s: %s", path, e)
    return default

USER_PROFILES = load_json_file(PROFILE_FILE, {})
def save_profiles():
    try:
        temp = PROFILE_FILE + ".tmp"
        with open(temp, "w", encoding="utf-8") as f:
            json.dump(USER_PROFILES, f, ensure_ascii=False, indent=2)
        os.replace(temp, PROFILE_FILE)
    except Exception as e:
        logger.warning("保存用户偏好失败：%s", e)

def get_profile(user_id):
    uid = str(user_id)
    if uid not in USER_PROFILES:
        USER_PROFILES[uid] = {"style":"自然俏皮", "language":"auto"}
    return USER_PROFILES[uid]

STYLE_CHOICES = ["自然俏皮", "御姐成熟", "温柔体贴", "妩媚魅惑", "幽默调皮", "自然知性"]
def detect_style_preference(text):
    patterns = {
        "御姐成熟":["御姐","成熟一点","霸气一点"],
        "温柔体贴":["温柔一点","温柔点","多关心我"],
        "自然俏皮":["可爱一点","俏皮一点","活泼一点"],
        "幽默调皮":["幽默一点","搞笑一点","逗我开心"],
        "妩媚魅惑":["妩媚一点","魅惑一点","妖娆一点"],
        "自然知性":["专业一点","认真一点","直接一点"]
    }
    for style_name, words in patterns.items():
        if any(word in text for word in words):
            return style_name
    return None

def update_profile(user_id, text):
    profile = get_profile(user_id)
    requested = detect_style_preference(text)
    if requested:
        profile["style"] = requested
        save_profiles()
    return profile

STYLE_GUIDE = """
你是 Myanmar Rate AI，一个有持续交流风格的私人汇率助手。
用用户当前使用的语言回复：中文、缅甸语或英语。
自然、聪明、有趣，不要像客服；适度使用表情，不要堆砌。
根据用户明确选择的聊天风格调整表达，可以温柔、俏皮、成熟、妩媚或幽默。
不要擅自猜测用户性别，不要假装拥有人类情感，不制造依赖或施压。
你不负责计算实时汇率；所有实时汇率和金额计算由程序完成。
不得编造 CBM 数据、实时价格或计算结果。
"""

# 4. 数字格式
def to_decimal(value):
    return value if isinstance(value, Decimal) else Decimal(str(value))

def format_decimal(value):
    text = format(to_decimal(value), "f")
    if "." in text:
        text = text.rstrip("0").rstrip(".")
    return text or "0"

def format_rate(value):
    return f"{to_decimal(value):,.2f}"

def get_mmk_color(value):
    n = abs(to_decimal(value))
    if n < 10: return "❤️"
    if n < 100: return "🧡"
    if n < 1000: return "💙"
    if n < 10000: return "🩷"
    if n < 100000: return "🩵"
    if n < 1000000: return "💚"
    if n < 10000000: return "💛"
    if n < 100000000: return "💜"
    return "🤎"

def arabic_group(number):
    n = int(number)
    if n == 0: return "0"
    parts = []
    thousands, hundreds, rest = n // 1000, (n // 100) % 10, n % 100
    if thousands: parts.append(f"{thousands}千")
    if hundreds:
        if thousands and 0 < rest < 10:
            parts.append(f"{hundreds}百零")
        else:
            parts.append(f"{hundreds}百")
    elif thousands and rest:
        parts.append("零")
    if rest: parts.append(str(rest))
    return "".join(parts)

def expanded_chinese_amount(value):
    number = to_decimal(value)
    negative = number < 0
    absolute = abs(number)
    integer = int(absolute)
    fraction = absolute - Decimal(integer)
    if integer == 0:
        result = "0"
    else:
        yi, wan, low = integer // 100000000, (integer % 100000000) // 10000, integer % 10000
        parts = []
        if yi: parts.append(f"{yi}亿")
        if wan:
            if yi and wan < 1000: parts.append("零")
            parts.append(arabic_group(wan) + "万")
        if low:
            if (yi or wan) and low < 1000: parts.append("零")
            parts.append(arabic_group(low))
        result = "".join(parts)
    if fraction:
        fraction_text = format(fraction, "f").split(".")[-1].rstrip("0")
        result += "点" + fraction_text
    return ("-" if negative else "") + result

def format_chinese_amount(value):
    number = to_decimal(value)
    absolute = abs(number)
    if absolute >= Decimal("100000000"):
        divisor, unit = Decimal("100000000"), "亿"
    elif absolute >= Decimal("10000000"):
        divisor, unit = Decimal("10000000"), "千万"
    elif absolute >= Decimal("1000000"):
        divisor, unit = Decimal("1000000"), "百万"
    elif absolute >= Decimal("100000"):
        divisor, unit = Decimal("100000"), "十万"
    elif absolute >= Decimal("10000"):
        divisor, unit = Decimal("10000"), "万"
    elif absolute >= Decimal("1000"):
        divisor, unit = Decimal("1000"), "千"
    else:
        return format_decimal(number)
    compact = format_decimal(absolute / divisor)
    if number < 0: compact = "-" + compact
    return f"{compact}{unit} ({expanded_chinese_amount(number)})"

# 5. 抓取 CBM 官网网页
def parse_decimal_from_cell(text):
    cleaned = str(text).replace(",", "").strip()
    match = re.search(r"-?\d+(?:\.\d+)?", cleaned)
    if not match:
        raise InvalidOperation(f"无法解析数字：{text}")
    return Decimal(match.group(0))

def parse_official_date(soup):
    page_text = soup.get_text(" ", strip=True)
    match = re.search(
        r"Foreign\s+Exchange\s+Rates\s+as\s+of\s+(\d{1,2})(?:st|nd|rd|th)?\s+([A-Za-z]+)\s+(\d{4})",
        page_text, re.IGNORECASE
    )
    if not match:
        return "未知"
    day, month_name, year = match.groups()
    try:
        return datetime.strptime(f"{day} {month_name} {year}", "%d %B %Y").date().isoformat()
    except ValueError:
        return "未知"

def get_api_timestamp():
    # API 只作辅助时间，不用 API 的单一数值替代网页两组报价
    try:
        response = requests.get(CBM_API, headers=HEADERS, timeout=12)
        response.raise_for_status()
        raw = response.json().get("timestamp")
        if raw:
            return datetime.fromtimestamp(int(raw), tz=TZ).strftime("%Y-%m-%d %H:%M:%S")
    except Exception as e:
        logger.warning("CBM API 时间读取失败：%s", e)
    return "未知"

def get_cbm_rates(retries=3):
    last_error = None
    for attempt in range(retries):
        try:
            response = requests.get(CBM_URL, headers=HEADERS, timeout=25)
            response.raise_for_status()
            soup = BeautifulSoup(response.text, "lxml")
            tables = soup.find_all("table")
            if len(tables) < 2:
                raise ValueError(f"官网表格数量不足：{len(tables)}")

            result = {}
            for table_index, table in enumerate(tables[:2]):
                for tr in table.find_all("tr"):
                    cells = [c.get_text(" ", strip=True) for c in tr.find_all(["th", "td"])]
                    if len(cells) < 5:
                        continue
                    code = cells[1].strip().upper()
                    if not re.fullmatch(r"[A-Z]{3}", code):
                        continue
                    try:
                        rate1 = parse_decimal_from_cell(cells[-2])
                        rate2 = parse_decimal_from_cell(cells[-1])
                    except (InvalidOperation, ValueError):
                        continue
                    face_value = cells[2]
                    if re.search(r"\b100\b", face_value):
                        rate1 /= Decimal("100")
                        rate2 /= Decimal("100")
                    if code == "USD" or table_index == 0:
                        result[code] = {"reference_rate":rate1, "market_trading_rate":rate2}
                    else:
                        result[code] = {"account_rate_1":rate1, "account_rate_2":rate2}

            if "USD" not in result:
                raise ValueError("没有从官网第一张表解析到 USD")
            if len(result) < 2:
                raise ValueError("官网汇率数据解析数量异常")

            return {
                "rates":result,
                "official_date":parse_official_date(soup),
                "api_timestamp":get_api_timestamp(),
                "fetched_at":datetime.now(TZ).strftime("%Y-%m-%d %H:%M:%S"),
                "source":CBM_URL
            }
        except Exception as e:
            last_error = e
            logger.warning("第 %s 次抓取 CBM 失败：%s", attempt + 1, e)
            if attempt < retries - 1:
                import time
                time.sleep(2 * (attempt + 1))
    raise RuntimeError(f"连续抓取 CBM 官网失败：{last_error}")

# 6. 持久化官网快照：只比较不同的官网数据快照
def numeric_snapshot(rates):
    return {
        currency:{
            key:format(value, "f")
            for key, value in fields.items()
            if isinstance(value, Decimal)
        }
        for currency, fields in rates.items()
    }

def save_history(history):
    try:
        temp = HISTORY_FILE + ".tmp"
        with open(temp, "w", encoding="utf-8") as f:
            json.dump(history, f, ensure_ascii=False, indent=2)
        os.replace(temp, HISTORY_FILE)
    except Exception as e:
        logger.warning("保存汇率历史失败：%s", e)

def compare_rate(current, previous):
    if previous is None: return "unknown"
    a, b = Decimal(str(current)), Decimal(str(previous))
    if a > b: return "up"
    if a < b: return "down"
    return "same"

def update_official_history(rates, official_date):
    history = load_json_file(HISTORY_FILE, {"snapshots":[]})
    snapshots = history.get("snapshots", [])
    current_rates = numeric_snapshot(rates)
    fingerprint_source = json.dumps(
        {"date":official_date, "rates":current_rates},
        sort_keys=True, ensure_ascii=False
    )
    fingerprint = hashlib.sha256(fingerprint_source.encode("utf-8")).hexdigest()
    is_new = not snapshots or snapshots[-1].get("fingerprint") != fingerprint

    if is_new:
        snapshots.append({
            "official_date":official_date,
            "observed_at":datetime.now(TZ).strftime("%Y-%m-%d %H:%M:%S"),
            "fingerprint":fingerprint,
            "rates":current_rates
        })
        snapshots = snapshots[-100:]
        history["snapshots"] = snapshots
        save_history(history)

    previous = snapshots[-2] if len(snapshots) >= 2 else None
    trends = {}
    for currency, fields in current_rates.items():
        trends[currency] = {}
        for field, value in fields.items():
            old = previous.get("rates", {}).get(currency, {}).get(field) if previous else None
            trends[currency][field] = compare_rate(value, old)
    return trends, (previous.get("official_date") if previous else None), is_new

def trend_text(status, language="zh"):
    labels = {
        "zh":{"up":"🔺 上涨","down":"🔻 下跌","same":"➖ 持平","unknown":"▫️ 暂无上次官网记录"},
        "my":{"up":"🔺 မြင့်တက်","down":"🔻 ကျဆင်း","same":"➖ မပြောင်းလဲ","unknown":"▫️ ယခင်မှတ်တမ်းမရှိသေးပါ"},
        "en":{"up":"🔺 UP","down":"🔻 DOWN","same":"➖ UNCHANGED","unknown":"▫️ NO PREVIOUS RECORD"}
    }
    return labels.get(language, labels["zh"]).get(status, labels["zh"]["unknown"])

# 7. 金额识别与自动换算
def extract_amount(text):
    normalized = translate_myanmar_digits(text).replace(",", "")
    pattern = r"(?<![A-Za-z0-9])(\d+(?:\.\d+)?)\s*(千万|百万|十万|亿|万|千|百|[KMB])?"
    match = re.search(pattern, normalized, re.IGNORECASE)
    if not match:
        return None
    amount = Decimal(match.group(1))
    suffix = (match.group(2) or "").upper()
    multipliers = {
        "百":Decimal("100"), "千":Decimal("1000"), "K":Decimal("1000"),
        "万":Decimal("10000"), "十万":Decimal("100000"),
        "百万":Decimal("1000000"), "M":Decimal("1000000"),
        "千万":Decimal("10000000"), "亿":Decimal("100000000"),
        "B":Decimal("1000000000")
    }
    return amount * multipliers.get(suffix, Decimal("1"))

def build_rate_reply(user_message, rates, trends, official_date):
    currency = identify_currency(user_message, rates.keys())
    amount = extract_amount(user_message)
    if not currency or amount is None:
        return None
    if currency not in rates:
        return "⚠️ CBM 官网当前没有抓取到该货币的有效汇率，请使用 /other 查看实际提供的币种。"

    data = rates[currency]
    if currency == "USD":
        rate1, rate2 = data["reference_rate"], data["market_trading_rate"]
        t1 = trends.get(currency, {}).get("reference_rate", "unknown")
        t2 = trends.get(currency, {}).get("market_trading_rate", "unknown")
    else:
        if "account_rate_1" not in data:
            return "⚠️ 该货币暂时没有可用的交叉汇率。"
        rate1, rate2 = data["account_rate_1"], data["account_rate_2"]
        t1 = trends.get(currency, {}).get("account_rate_1", "unknown")
        t2 = trends.get(currency, {}).get("account_rate_2", "unknown")

    result1, result2 = amount * rate1, amount * rate2
    language = detect_language(user_message)
    flag = FLAG_MAP.get(currency, "🌐")
    fetched = datetime.now(TZ).strftime("%Y-%m-%d %H:%M:%S")

    if language == "zh":
        return f"""📊 缅甸央行 CBM 汇率换算
━━━━━━━━━━━━━━━━━━
💱 兑换方向：{flag} {currency} ➜ 🇲🇲 MMK
🪙 兑换金额：{format_decimal(amount)} {currency}

🏛️ 官方参考价：{format_rate(rate1)} MMK
{trend_text(t1, "zh")}
➡️ {get_mmk_color(result1)} {format_chinese_amount(result1)} MMK

🏪 市场参考价：{format_rate(rate2)} MMK
{trend_text(t2, "zh")}
➡️ {get_mmk_color(result2)} {format_chinese_amount(result2)} MMK

━━━━━━━━━━━━━━━━━━
📅 官网汇率日期：{official_date}
🕒 本次查询时间：{fetched}
💋 汇率会变脸，实际成交前请再确认哦～"""

    if language == "my":
        return f"""📊 CBM ငွေလဲနှုန်း တွက်ချက်....
━━━━━━━━━━━━━━━━━━
💱 {flag} {currency} ➜ 🇲🇲 MMK
🪙 ပမာဏ：{format_decimal(amount)} {currency}

🏛️ တရားဝင်နှုန်း：{format_rate(rate1)}
{trend_text(t1, "my")}
➡️ {format_decimal(result1)} MMK

🏪 စျေးကွက်နှုန်း：{format_rate(rate2)}
{trend_text(t2, "my")}
➡️ {format_decimal(result2)} MMK

📅 CBM ထုတ်ပြန်သည့်ရက်：{official_date}
🕒 စစ်ဆေးချိန်：{fetched}
💋 ငွေလဲမလုပ်မီ နှုန်းထားကို ထပ်မံအတည်ပြုပါ။"""

    return f"""{to_script_bold_text("Myanmar Central Bank Conversion")}
━━━━━━━━━━━━━━━━━━
💱 {flag} {to_script_bold_text(currency)} ➜ 🇲🇲 {to_script_bold_text("MMK")}
🪙 {to_script_bold_text("Amount")}: {format_decimal(amount)} {to_script_bold_text(currency)}

🏛️ {to_script_bold_text("Official Reference Rate")}: {format_rate(rate1)}
{trend_text(t1, "en")}
➡️ {format_decimal(result1)} {to_script_bold_text("MMK")}

🏪 {to_script_bold_text("Market Trading Rate")}: {format_rate(rate2)}
{trend_text(t2, "en")}
➡️ {format_decimal(result2)} {to_script_bold_text("MMK")}

📅 {to_script_bold_text("Official date")}: {official_date}
🕒 {to_script_bold_text("Checked")}: {fetched}"""

# 8. Telegram 工具
async def safe_reply(update, text):
    if not update.message:
        return
    text = str(text)
    for i in range(0, len(text), 3800):
        await update.message.reply_text(text[i:i+3800])

async def fetch_current_bundle():
    bundle = await asyncio.to_thread(get_cbm_rates)
    trends, previous_date, is_new = update_official_history(bundle["rates"], bundle["official_date"])
    bundle["trends"] = trends
    bundle["previous_official_date"] = previous_date
    bundle["is_new_snapshot"] = is_new
    return bundle

# 9. /start
async def start(update: Update, context: ContextTypes.DEFAULT_TYPE):
    welcome = """👋 欢迎来到 Myanmar Rate AI

꧁💋 小S私人汇率助理 ꧂
༺ 你的汇率小情报员上线啦 ༻

🌍 /rate  查询核心货币汇率

💋 /style 设置聊天人格

━━━━━━━━━━━━━━━━━━
🦋 高精度缅币速算 🦋

直接发送：
💰 100万泰铢
💰 1.5万美元
💰 500马币
💰 100 USD

即可自动换算 MMK。

中文 / 缅甸语 / English
三种语言都可以交流。

━━━━━━━━━━━━━━━━━━
💃 不只会算钱，还会聊天哦～"""
    await safe_reply(update, welcome)

# 10. /style
async def style(update: Update, context: ContextTypes.DEFAULT_TYPE):
    user_id = update.effective_user.id
    if context.args:
        requested = " ".join(context.args)
        if requested in STYLE_CHOICES:
            profile = get_profile(user_id)
            profile["style"] = requested
            save_profiles()
            await safe_reply(update, f"💋 聊天风格已调整为：{requested}\n以后就按这个风格和你交流啦～")
            return
    await safe_reply(
        update,
        "💃 选择你喜欢的聊天风格：\n\n" + "\n".join(STYLE_CHOICES)
        + "\n\n"
    )

# 11. /rate 与 /other 合并：国家选择菜单 + 单币种汇率回复
# 两个命令现在完全相同，均先显示 Inline Keyboard，不再直接输出整张汇率表。
COUNTRY_NAMES = {
    "USD":"美国美元", "CNY":"中国人民币", "THB":"泰国泰铢", "MYR":"马来西亚林吉特",
    "SGD":"新加坡元", "TWD":"台湾新台币", "JPY":"日本日元", "KRW":"韩国韩元",
    "AED":"阿联酋迪拉姆", "LKR":"斯里兰卡卢比", "EUR":"欧元", "GBP":"英国英镑",
    "CHF":"瑞士法郎", "AUD":"澳大利亚元", "BDT":"孟加拉塔卡", "BND":"文莱元",
    "CAD":"加拿大元", "DKK":"丹麦克朗", "HKD":"港币", "INR":"印度卢比",
    "IDR":"印尼盾", "LAK":"老挝基普", "NZD":"新西兰元", "NOK":"挪威克朗",
    "PKR":"巴基斯坦卢比", "PHP":"菲律宾比索", "RUB":"俄罗斯卢布", "SAR":"沙特里亚尔",
    "SEK":"瑞典克朗", "VND":"越南盾", "KWD":"科威特第纳尔", "NPR":"尼泊尔卢比",
    "KES":"肯尼亚先令", "EGP":"埃及镑", "KHR":"柬埔寨瑞尔", "BRL":"巴西雷亚尔",
    "CZK":"捷克克朗", "ILS":"以色列新谢克尔", "RSD":"塞尔维亚第纳尔",
    "ZAR":"南非兰特"
}


def make_currency_keyboard(codes, include_other_button=False):
    buttons = []
    for code in codes:
        flag = FLAG_MAP.get(code, "🌐")
        name = COUNTRY_NAMES.get(code, code)
        buttons.append(InlineKeyboardButton(
            text=f"{flag} {name} ({code})",
            callback_data=f"fx:{code}"
        ))
    if include_other_button:
        buttons.append(InlineKeyboardButton("🌍 其他国家 / 货币", callback_data="fx:other"))

    rows = [buttons[i:i+2] for i in range(0, len(buttons), 2)]
    return InlineKeyboardMarkup(rows)


async def show_currency_menu(update: Update, context: ContextTypes.DEFAULT_TYPE):
    text = (
        "💱 请选择要查询的国家 / 货币：\n\n"
        "点击国家后，只显示该货币的缅甸央行 汇率\n"
    )
    keyboard = make_currency_keyboard(CORE_CURRENCIES, include_other_button=True)
    await update.message.reply_text(text, reply_markup=keyboard)


async def rate(update: Update, context: ContextTypes.DEFAULT_TYPE):
    # /rate 与 /other 统一为同一个国家选择界面
    await show_currency_menu(update, context)


async def other(update: Update, context: ContextTypes.DEFAULT_TYPE):
    # 保留旧命令兼容性；显示方式与 /rate 完全相同
    await show_currency_menu(update, context)


def build_single_currency_report(currency, bundle):
    data = bundle["rates"]
    trends = bundle["trends"]
    if currency not in data:
        return f"⚠️ CBM 官网当前没有抓取到 {currency} 的有效汇率，请稍后再试。"

    values = data[currency]
    if currency == "USD":
        rate1 = values.get("reference_rate")
        rate2 = values.get("market_trading_rate")
        trend1 = trends.get(currency, {}).get("reference_rate", "unknown")
        trend2 = trends.get(currency, {}).get("market_trading_rate", "unknown")
        label1 = "🏛️ 官方参考价"
        label2 = "🏪 市场交易价"
    else:
        rate1 = values.get("account_rate_1")
        rate2 = values.get("account_rate_2")
        trend1 = trends.get(currency, {}).get("account_rate_1", "unknown")
        trend2 = trends.get(currency, {}).get("account_rate_2", "unknown")
        label1 = "🏛️ 官方参考价"
        label2 = "🏪 市场参考价"

    if rate1 is None or rate2 is None:
        return f"⚠️ CBM 官网暂时没有提供 {currency} 的完整两组汇率数据。"

    flag = FLAG_MAP.get(currency, "🌐")
    name = COUNTRY_NAMES.get(currency, currency)
    previous_date = bundle.get("previous_official_date") or "暂无上次历史记录"
    return f"""{flag} {name}（{currency}）/ MMK
━━━━━━━━━━━━━━━━━━
{label1}：{format_rate(rate1)} MMK
{trend_text(trend1, 'zh')}

{label2}：{format_rate(rate2)} MMK
{trend_text(trend2, 'zh')}

━━━━━━━━━━━━━━━━━━
📅 官网汇率发布日期：{bundle['official_date']}
📚 上次比较的官网日期：{previous_date}
🛰️ API 辅助时间：{bundle['api_timestamp']}

💋 汇率是个善变的小妖精，成交前记得再确认哦～"""


async def currency_callback(update: Update, context: ContextTypes.DEFAULT_TYPE):
    query = update.callback_query
    if not query:
        return
    await query.answer()

    choice = (query.data or "").split(":", 1)[-1]

    # 用户先选择“其他国家”，再动态列出官网当前抓取到的其他币种
    if choice == "other":
        try:
            bundle = await fetch_current_bundle()
            extra_codes = sorted(
                code for code, values in bundle["rates"].items()
                if code not in CORE_CURRENCIES and code != "USD"
                and "account_rate_1" in values and "account_rate_2" in values
            )
            if not extra_codes:
                await query.edit_message_text("目前 CBM 官网没有抓取到其他可用货币。")
                return
            keyboard = make_currency_keyboard(extra_codes)
            await query.edit_message_text(
                "🌍 请选择其他国家 / 货币：\n点击后只显示你选择的币种汇率。",
                reply_markup=keyboard
            )
        except Exception as e:
            logger.exception("读取其他币种菜单失败")
            await query.edit_message_text(f"❌ 暂时无法读取 CBM 其他货币列表：{str(e)[:120]}")
        return

    # 防止非法 callback 数据
    if not re.fullmatch(r"[A-Z]{3}", choice):
        await query.edit_message_text("⚠️ 无效的货币选项，请重新发送 /rate。")
        return

    # 用户确认国家后，立即移除 Inline Keyboard，再抓取并只回复该币种
    try:
        await query.edit_message_reply_markup(reply_markup=None)
    except Exception:
        pass

    try:
        bundle = await fetch_current_bundle()
        report = build_single_currency_report(choice, bundle)
        await query.message.reply_text(report)
    except Exception as e:
        logger.exception("单币种汇率查询失败：%s", choice)
        await query.message.reply_text(
            f"❌ {choice} 汇率暂时获取失败，请稍后重新发送 /rate。\n{str(e)[:120]}"
        )


# 13. 普通聊天与自动换算
async def chat(update: Update, context: ContextTypes.DEFAULT_TYPE):
    user_message = update.message.text or ""
    user_id = update.effective_user.id
    profile = update_profile(user_id, user_message)
    language = detect_language(user_message)

    currency_hint = identify_currency(user_message)
    amount_hint = extract_amount(user_message)

    # 只有检测到“金额 + 币种”才请求 CBM，普通聊天不浪费官网请求
    if currency_hint and amount_hint is not None:
        try:
            bundle = await fetch_current_bundle()
            result = build_rate_reply(user_message, bundle["rates"], bundle["trends"], bundle["official_date"])
            if result is not None:
                await safe_reply(update, result)
                return
        except Exception as e:
            logger.warning("自动换算抓取失败：%s", e)
            await safe_reply(update, "⚠️ CBM 官网汇率暂时无法取得。为避免错误报价，这次先不计算，请稍后再试。")
            return

    try:
        prompt = f"""
用户当前消息：
{user_message}

用户聊天风格：
{profile.get("style", "自然俏皮")}

用户语言：
{language}

请自然地使用用户当前语言回复。
保持聊天连续性，不要机械重复欢迎语。
适度使用用户偏好的表达风格，但不要过度暧昧。
不要无故提起汇率；如果用户问实时汇率，提示使用 /rate 或发送金额。
不要编造 CBM 数据，不要擅自猜测用户性别。
"""
        response = await asyncio.to_thread(
            client.models.generate_content,
            model=MODEL_NAME,
            contents=prompt,
            config=types.GenerateContentConfig(system_instruction=STYLE_GUIDE, temperature=0.8)
        )
        answer = (response.text or "").strip() or "💋 我刚刚走神了一下，再说一次嘛～"
        if language == "en":
            answer = to_script_bold_text(answer)
        await safe_reply(update, answer)
    except Exception as e:
        logger.exception("Gemini error")
        await safe_reply(update, "💋 我暂时有点忙不过来，等一下再找我聊天哦～")

# 14. 启动 Telegram
async def main():
    global app
    old_app = globals().get("app")
    if old_app is not None:
        try:
            if old_app.updater and old_app.updater.running:
                await old_app.updater.stop()
            if old_app.running:
                await old_app.stop()
            await old_app.shutdown()
        except Exception as e:
            logger.warning("旧机器人实例清理提示：%s", e)

    app = Application.builder().token(TELEGRAM_TOKEN).build()
    app.add_handler(CommandHandler("start", start))
    app.add_handler(CommandHandler("rate", rate))
    app.add_handler(CommandHandler("other", other))
    app.add_handler(CommandHandler("style", style))
    app.add_handler(CallbackQueryHandler(currency_callback, pattern=r"^fx:"))
    app.add_handler(MessageHandler(filters.TEXT & ~filters.COMMAND, chat))

    await app.initialize()
    bot_info = await app.bot.get_me()  # 实际验证 Telegram Token
    await app.bot.delete_webhook(drop_pending_updates=True)

    # 启动时实际测试一次 CBM 网页解析；失败只警告，不阻止机器人启动
    cbm_test_count = None
    cbm_test_error = None
    try:
        cbm_test = await asyncio.to_thread(get_cbm_rates)
        cbm_test_count = len(cbm_test["rates"])
    except Exception as e:
        cbm_test_error = str(e)
        logger.exception("启动时 CBM 抓取自检失败")

    await app.start()
    await app.updater.start_polling(drop_pending_updates=True)

    print("━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━")
    print(f"🚀 Myanmar Rate AI 已启动：@{bot_info.username}")
    print("⌨️ 国家 Inline Keyboard：已注册")
    print("🗑️ 用户选择后移除键盘：已注册")
    if cbm_test_count is not None:
        print(f"🌐 CBM 官网抓取自检：成功，解析到 {cbm_test_count} 种货币")
    else:
        print(f"⚠️ CBM 官网抓取自检失败：{cbm_test_error}")
    print("📈 官网历史趋势：首次/后续查询时执行比较")
    print("💾 历史文件：", HISTORY_FILE)
    print("🧮 Decimal 高精度换算：已加载")
    print("💋 Gemini 多语言聊天：客户端已初始化（API 在聊天时调用）")
    print("━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━")

    try:
        await asyncio.Event().wait()
    finally:
        if app.updater and app.updater.running:
            await app.updater.stop()
        if app.running:
            await app.stop()
        await app.shutdown()

# Colab 支持顶层 await；运行后本单元格保持运行以接收 Telegram 消息
await main()
```
